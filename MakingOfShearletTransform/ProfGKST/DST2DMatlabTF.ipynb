```python
import numpy as np
# import matplotlib.pyplot as plt
import cv2
from scipy.fftpack import fftshift, ifftshift, fft2, ifft2
```

### sample input $x$, $DST(x)$ and $IDST(x)$ 

```python
# Load the image
x = cv2.imread('/data1/kishoretarafdar/DATASETS.port/lena.jpg')  # Replace with your image path
# Convert to grayscale
x = cv2.cvtColor(x, cv2.COLOR_BGR2GRAY)

# Resize to 256x256
x = cv2.resize(x, (256, 256))
x.shape
```

```python
x = x/255
```

### DST


     from codes of Prof G. Kutyniok

```python
#forward discrete shearlet transform
"""MATLAB Prof GK
   
    coeffs = zeros(size(shearletSystem.shearlets));
    
 
    %get data in frequency domain
    Xfreq = fftshift(fft2(ifftshift(X)));

    %compute shearlet coefficients at each scale
    %not that pointwise multiplication in the fourier domain equals convolution
    %in the time-domain
    for j = 1:shearletSystem.nShearlets
        coeffs(:,:,j) = fftshift(ifft2(ifftshift(Xfreq.*conj(shearletSystem.shearlets(:,:,j)))));        
    end

"""


## FT of X
xfreq = fftshift(fft2(ifftshift(x)))

## DST
coeffs = np.zeros_like(shearlets)
for j in range(1,shearlets.shape[-1]):
    coeffs[:,:,j] = fftshift(ifft2(ifftshift(xfreq * (shearlets[:,:,j]).conjugate())))
coeffs.shape, coeffs.dtype
```

Shearlet coefficients are real numbers $\mathbb{R}$. 

        Imaginary part of the  is ~0

```python
np.real(coeffs)
np.sum(np.imag(coeffs))
```

### IDST 

     from codes of Prof G. Kutyniok

```python
##inverse DST
"""MATLAB Prof GK
    
    X = zeros(size(coeffs,1),size(coeffs,2));


    for j = 1:shearletSystem.nShearlets
        X = X+fftshift(fft2(ifftshift(coeffs(:,:,j)))).*shearletSystem.shearlets(:,:,j);
    end
    X = fftshift(ifft2(ifftshift((1./shearletSystem.dualFrameWeights).*X)));
    

"""

xr = np.zeros_like(x)
for j in range(1,shearlets.shape[-1]):
    xr = xr+fftshift(fft2(ifftshift(coeffs[:,:,j]))) * shearlets[:,:,j]
xr = fftshift(ifft2(ifftshift((1/dualframeweights)*xr)))
# xr = fftshift(ifft2(ifftshift(xr)))
xr = np.real(xr)
xr.shape, np.linalg.norm(x), np.linalg.norm(xr)
```
